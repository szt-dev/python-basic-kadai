```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}さんは大人です")
        else:
            print(f"{self.name}さんは大人ではありません")


personal_data = {"田中": 25, "佐藤": 18, "鈴木": 20}

# ① インスタンスを作ってリストに入れる
humans = []
for name, age in personal_data.items():
    humans.append(Human(name, age))

# ② リストの中身に対してメソッドを呼ぶ
for human in humans:
    human.check_adult()
```
